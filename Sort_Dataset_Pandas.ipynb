```python
import pandas as pd
```

```python
data = {
    "Name": ["Aarav", "Priya", "Rahul", "Sneha", "Vikram", "Ananya", "Rohan", "Neha"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "HR"],
    "Age": [23, 25, 22, 28, 24, 21, 26, 23],
    "Marks": [85, 72, 91, 68, 79, 95, 74, 88],
    "Attendance": [92, 85, 96, 78, 88, 98, 82, 94]
}

df = pd.DataFrame(data)

df
```

```python
print("Number of rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns)
```

```python
marks_ascending = df.sort_values(by="Marks", ascending=True)

print("Students sorted by Marks in ascending order:")
marks_ascending
```

```python
marks_descending = df.sort_values(by="Marks", ascending=False)

print("Students sorted by Marks in descending order:")
marks_descending
```

```python
department_ascending = df.sort_values(by="Department", ascending=True)

print("Students sorted by Department in ascending order:")
department_ascending
```

```python
department_descending = df.sort_values(by="Department", ascending=False)

print("Students sorted by Department in descending order:")
department_descending
```

```python
multiple_sort = df.sort_values(
    by=["Department", "Marks"],
    ascending=[True, False]
)

print("Sorted by Department ascending and Marks descending:")
multiple_sort
```

```python
attendance_marks_sort = df.sort_values(
    by=["Attendance", "Marks"],
    ascending=[False, False]
)

print("Sorted by Attendance and Marks in descending order:")
attendance_marks_sort
```
